## 8. Explanatory Wrangling

a. Perform the following wrangling:

i. select the `State` column and the rows corresponding to water conflicts occurring

```python
# Step 1: Select State column where conflict is present
states_conflict = df.loc[df['Conflict Present'] == 'Y', 'State']

# Step 2: Split on semicolon delimiter
states_split = states_conflict.str.split(';')

# Step 3: Explode lists into individual rows
states_exploded = states_split.explode()

# Step 4: Unique values, keeping NAs
states_unique = states_exploded.unique()

states_unique

# Doing it all in one line:
#df.loc[df['Conflict Present'] == 'Y', 'State'].str.split(';').explode().unique()
```

## 10. Visualize data exploration & discussion

a. Create a bar plot showing the number of conflicts reported per state and write a description of these results. Include a bar for NA values.

```python
# Count conflicts per state, keeping NAs
state_counts = (
    df.loc[df['Conflict Present'] == 'Y', 'State']
    .str.split(';')
    .explode()
    .str.strip()
    .value_counts(dropna=False)
)

# Plot
state_counts.plot(
    kind='bar',
    xlabel='State',
    ylabel='Number of conflicts',
    title='Water conflicts reported per state in the Colorado River (2005-2021)'
);
```
